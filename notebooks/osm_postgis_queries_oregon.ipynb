### 📚 Step 2: Import Required Libraries

We will use the following tools:

- `geopandas`: to read spatial query results into GeoDataFrames and visualize them
- `sqlalchemy`: to create a database connection engine for PostGIS
- `matplotlib.pyplot`: to visualize the results
- `pathlib`: to work with file paths more cleanly

````python
import geopandas as gpd
import matplotlib.pyplot as plt
from sqlalchemy import create_engine
from pathlib import Path

print("Libraries imported!")
````

### 🔌 Step 3: Connect to the PostGIS Database

Before running SQL queries, create a connection to the database using SQLAlchemy.

**💡 This engine will be used throughout the notebook for all queries and data access!**

````python
# Create a SQLAlchemy engine to connect to the PostGIS database
engine = create_engine(
    "postgresql+psycopg2://postgres:postgres@localhost:5432/oregon"
)
print("SQLAlchemy engine created!")
````

### ▶️ Step 4: Run Query 1 - Library Distribution

This query extracts library point locations for spatial distribution analysis.

The SQL query is stored in a separate `.sql` file. In this step, we read the query into Python, send it to the PostGIS database, and load the result as a GeoDataFrame.

````python
query_1_file = Path("../sql/oregon/01_osm_library_distribution.sql")

# Read SQL query from file
query_1_sql = query_1_file.read_text(encoding="utf-8")

# Execute query and load result as a GeoDataFrame
try:
    query_1_results = gpd.read_postgis(query_1_sql, engine, geom_col="geom")
    # Display results
    display(query_1_results)
except Exception as e:
    print("❌ Query failed:")
    print(e)
````

### 🗺️ Step 5: Visualize Query 1 Results

Now that the query results have been loaded as a GeoDataFrame, we can create a map of all of the library locations in the state. 

````python
# Extract coordinates
x = query_1_results.geometry.x
y = query_1_results.geometry.y

fig, ax = plt.subplots(figsize=(12, 8))

# Library locations
sp = plt.scatter(
    x, y,
    color = 'turquoise',
    s = 25
)
# add labels
ax.set_title("Library Locations in Oregon", fontsize=16)
ax.set_axis_off()

plt.tight_layout()
plt.show()
````

### Query 1 Reflection
This map shows the location of all libraries within the state of Oregon. They are distributed throughout the state, but the most concentrated area is around Portland, the most populated city in the state.

### ▶️ Step 6: Run Query 2 - Railroad Length by County

This query calculates the total length of railways in each county (including rails, light rails, and trams) in kilometers.

````python
query_2_file = Path("../sql/oregon/02_osm_railroad_length_by_county.sql")

# Read SQL query from file
query_2_sql = query_2_file.read_text(encoding="utf-8")

# Execute query and load result as a GeoDataFrame
try:
    query_2_results = gpd.read_postgis(query_2_sql, engine, geom_col="geom")
    # Display results
    display(query_2_results)
except Exception as e:
    print("❌ Query failed:")
    print(e)
````

### 🗺️ Step 7: Visualize Query 2 Results as a Choropleth Map

Now that the query results have been loaded as a GeoDataFrame, we can create a choropleth map.

In this map, each county is colored by its **total railroad length (in kilometers)**.

````python
viz_column = "railroad_length_km"

# Select columns for the table
table_df = (
    query_2_results[["county_name", viz_column]]
    .sort_values(viz_column, ascending=False)
    .copy()
)

table_df[viz_column] = table_df[viz_column].round(2)

fig, (ax_map, ax_table) = plt.subplots(
    1, 2,
    figsize=(16, 8),
    gridspec_kw={"width_ratios": [3, 1]}
)

# Map
query_2_results["railroad_length_km"] = (
    query_2_results["railroad_length_km"].fillna(0)
)

query_2_results.plot(
    column=viz_column,
    legend=True,
    ax=ax_map,
    vmin=query_2_results[viz_column].min(),
    vmax=query_2_results[viz_column].max(),
    cmap="GnBu"
)

ax_map.set_title("Railroad Length by County (km)", fontsize=16)
ax_map.set_axis_off()

# Table
ax_table.axis("off")
tbl = ax_table.table(
    cellText=table_df.values,
    colLabels=["County", "Railroad Length (km)"],
    loc="center",
    cellLoc="center"
)

tbl.auto_set_font_size(False)
tbl.set_fontsize(10)
tbl.scale(1, 1.5)

plt.tight_layout()
plt.show()
````

### Query 2 Reflection
This map shows the total railroad length by county as a choropleth map. Many counties on the East side on the state have no railways or very short distances of railways (less than 100km). The counties on the West side of the state generally have longer railway distances, with Portland having the longest distance. This makes sense as the West side of the state is generally more populated and Portland is a large city with public transportation infrastructure.

### ▶️ Step 8: Run Query 3 - Restaurants Near Streets

This query identifies streets (cycleways) with the highest number of restaurants located within 0.25 miles.

````python
query_3_file = Path("../sql/oregon/03_osm_library_density_by_county.sql")

# Read SQL query from file
query_3_sql = query_3_file.read_text(encoding="utf-8")

# Execute query and load result as a GeoDataFrame
try:
    query_3_results = gpd.read_postgis(query_3_sql, engine, geom_col="geom")
    # Display results
    display(query_3_results)
except Exception as e:
    print("❌ Query failed:")
    print(e)
````

### 📊 Step 9: Visualize Query 3 Results as a Choropleth Map

This map shows the density of libraries by county as number of libraries per sq km. 

````python
viz_column = "library_density_per_sq_km"

# Map
query_3_results["library_density_per_sq_km"] = (
    query_3_results["library_density_per_sq_km"].fillna(0)
)

fig, (ax_map) = plt.subplots(
    figsize=(16, 8),
)

query_3_results.plot(
    column=viz_column,
    legend=True,
    ax=ax_map,
    vmin=query_3_results[viz_column].min(),
    vmax=query_3_results[viz_column].max(),
    cmap="RdPu"
)

ax_map.set_title("Library Density by County (libraries/sq km)", fontsize=16)
ax_map.set_axis_off()

# Table
ax_table.axis("off")
tbl = ax_table.table(
    cellText=table_df.values,
    colLabels=["County", "Library density (lib/sq km)"],
    loc="center",
    cellLoc="center"
)

tbl.auto_set_font_size(False)
tbl.set_fontsize(10)
tbl.scale(1, 1.5)

plt.tight_layout()
plt.show()
````

### Query 3 Reflection
This choropleth map shows the density of libraries in each county as libraries per square kilometer. Again, the highest density is in Multnomah county which countains Portland, the most populous city in the state. Many of the counties in the Eastern half of the state have very low library density.
